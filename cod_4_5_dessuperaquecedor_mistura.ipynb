```python
!pip install CoolProp -q
import CoolProp.CoolProp as CP

P = 10e5  # Pa (10 bar)
m1 = 5.0  # kg/s

h1 = CP.PropsSI('H', 'P', P, 'T', 350.0 + 273.15, 'Water')
h2 = CP.PropsSI('H', 'P', P, 'T', 40.0 + 273.15, 'Water')
h3 = CP.PropsSI('H', 'P', P, 'Q', 1.0, 'Water')

m2 = m1 * (h1 - h3) / (h3 - h2)
m3 = m1 + m2

print(f"Vazao de agua liquida injetada (m2): {m2:.4f} kg/s")
print(f"Vazao total de vapor saturado  (m3): {m3:.4f} kg/s")
```
