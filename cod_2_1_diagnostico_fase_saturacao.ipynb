```python
!pip install CoolProp -q
from CoolProp.CoolProp import PropsSI, PhaseSI

# Definição dos parâmetros operacionais (sempre no Sistema Internacional)
fluido = 'R134a'
P_sistema = 300000.0  # Pressão absoluta em Pa (3 bar)

# 1. Temperatura de saturação (independe se Q=0 ou Q=1 dentro do envelope de fases)
T_sat = PropsSI('T', 'P', P_sistema, 'Q', 0, fluido)

# 2. Entalpias especificas nos limites do envelope de saturação (J/kg)
hf = PropsSI('H', 'P', P_sistema, 'Q', 0, fluido)  # Liquido saturado (x = 0)
hg = PropsSI('H', 'P', P_sistema, 'Q', 1, fluido)  # Vapor saturado (x = 1)
h_fg = hg - hf  # Calor latente de vaporização

# 3. Diagnostico automático de fase a 50 oC (323.15 K) e 3 bar
fase = PhaseSI('P', P_sistema, 'T', 323.15, fluido)

print(f"Temperatura de saturação ({fluido} a 3 bar): {T_sat - 273.15:.2f} ºC")
print(f"Calor latente de vaporização (h_fg): {h_fg / 1000:.2f} kJ/kg")
print(f"Estado de agregação a 50 ºC e 3 bar: {fase}")
```
