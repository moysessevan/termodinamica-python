```python
!pip install CoolProp --quiet
import CoolProp.CoolProp as CP

P1 = 7e5       # Pa
T1 = 300.0     # K
V1 = 0.10      # m3
V2 = 0.20      # m3

D1 = CP.PropsSI('D', 'P', P1, 'T', T1, 'Nitrogen')
u1 = CP.PropsSI('U', 'P', P1, 'T', T1, 'Nitrogen')
m = D1 * V1
D2 = m / V2

# Caso (a): Expansao livre no vacuo -> u2 = u1
T2_vacuo = CP.PropsSI('T', 'D', D2, 'U', u1, 'Nitrogen')

# Caso (b): Expansao contra atmosfera -> dU = -P_atm * (V2 - V1)
P_atm = 101300.0
W_atm = -P_atm * (V2 - V1)
u2_atm = u1 + W_atm / m
T2_atm = CP.PropsSI('T', 'D', D2, 'U', u2_atm, 'Nitrogen')

print(f"Massa de N2: {m:.4f} kg")
print(f"Caso (a) Vacuo:     T2 = {T2_vacuo:.2f} K (dT = {T2_vacuo - T1:.2f} K)")
print(f"Caso (b) Atmosfera: T2 = {T2_atm:.2f} K (dT = {T2_atm - T1:.2f} K)")
```
