```python
import numpy as np
import matplotlib.pyplot as plt

P1 = 200e3   # Pa
V1 = 0.04    # m3
V2 = 0.12    # m3
V = np.linspace(V1, V2, 200)

expoentes = [0.0, 1.0, 1.3, 1.4]
nomes = ['Isobarico (n=0)', 'Isotermico (n=1)',
         'Politropico (n=1.3)', 'Adiabatico (n=1.4)']

plt.figure(figsize=(8, 5))

for n, nome in zip(expoentes, nomes):
    if n == 1.0:
        P = (P1 * V1) / V
        Wb = -P1 * V1 * np.log(V2 / V1)
    else:
        C = P1 * (V1 ** n)
        P = C / (V ** n)
        P2 = C / (V2 ** n)
        Wb = (P2 * V2 - P1 * V1) / (n - 1.0)
    print(f"{nome:22s}: W_b = {Wb/1000:7.2f} kJ")
    plt.plot(V, P / 1000.0, linewidth=2, label=f"{nome} | W={Wb/1000:.2f} kJ")

plt.fill_between(V, (P1 * V1) / V / 1000.0, color='skyblue', alpha=0.25,
                 label='Area sob n=1 (|Wb|)')
plt.title('Trabalho de Fronteira para Diferentes Trajetorias (P-V)')
plt.xlabel('Volume V (m3)')
plt.ylabel('Pressao P (kPa)')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.7)
plt.tight_layout()
plt.show()
```
