```python
# Definindo uma funcao propria para conversao de temperatura
def celsius_para_kelvin(temp_c):
    "Recebe temperatura em graus Celsius e retorna em Kelvin."
    return round (temp_c + 273.15,2)

# Lista de temperaturas medidas ao longo de um tunel de congelamento (em C)
temperaturas_processo_C = [25.0, 15.0, 4.0, -5.0, -18.0, -30.0]

# Criando uma lista vazia para armazenar os resultados
temperaturas_processo_K = []

# Estrutura de repeticao (loop 'for') - note o recuo (indentacao) de 4 espacos
for t in temperaturas_processo_C:
    t_k = celsius_para_kelvin(t)
    temperaturas_processo_K.append(t_k)

print("Temperaturas em C:", temperaturas_processo_C)
print("Temperaturas em K:", temperaturas_processo_K)
```
