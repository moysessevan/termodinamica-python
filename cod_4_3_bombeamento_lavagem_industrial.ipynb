```python
!pip install CoolProp -q
import CoolProp.CoolProp as CP

V_dot = 10.0 / 3600.0   # m3/s
T = 25.0 + 273.15       # K
P1, P2 = 1e5, 4.5e5     # Pa
z1, z2 = 0.0, 12.0      # m
g = 9.80665             # m/s2
eta_B = 0.75

rho = CP.PropsSI('D', 'T', T, 'P', P1, 'Water')
m_dot = rho * V_dot

# Potencia hidraulica minima (ideal) e potencia real
W_min = m_dot * ((P2 - P1) / rho + g * (z2 - z1))
W_real = W_min / eta_B
W_hp = W_real / 745.7   # 1 hp = 745.7 W

print(f"Densidade da agua: {rho:.2f} kg/m3 | m_dot = {m_dot:.3f} kg/s")
print(f"Potencia ideal:    {W_min/1000:.3f} kW")
print(f"Potencia real:     {W_real/1000:.3f} kW ({W_hp:.2f} hp)")
```
