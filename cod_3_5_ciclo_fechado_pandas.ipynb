```python
import pandas as pd

Ru = 8.314462618  # J/(mol.K)
Cv = 1.5 * Ru
Cp = 2.5 * Ru

P1, T1 = 1e5, 300.0
P2 = 2e5
P3 = 1e5

# Etapa 1 -> 2: Isocorica (V2 = V1 -> T2/T1 = P2/P1)
T2 = T1 * (P2 / P1)
W_12 = 0.0
dU_12 = Cv * (T2 - T1)
dH_12 = Cp * (T2 - T1)
Q_12 = dU_12

# Etapa 2 -> 3: Adiabatica reversivel (Q = 0)
T3 = T2 * (P3 / P2) ** (Ru / Cp)
Q_23 = 0.0
dU_23 = Cv * (T3 - T2)
dH_23 = Cp * (T3 - T2)
W_23 = dU_23

# Etapa 3 -> 1: Isobarica
dU_31 = Cv * (T1 - T3)
dH_31 = Cp * (T1 - T3)
Q_31 = dH_31
W_31 = dU_31 - Q_31

tabela = pd.DataFrame({
    'Etapa': ['1->2 (Isocorica)', '2->3 (Adiabatica)', '3->1 (Isobarica)', 'CICLO TOTAL'],
    'Q (J/mol)':  [Q_12, Q_23, Q_31, Q_12 + Q_23 + Q_31],
    'W (J/mol)':  [W_12, W_23, W_31, W_12 + W_23 + W_31],
    'dU (J/mol)': [dU_12, dU_23, dU_31, dU_12 + dU_23 + dU_31],
    'dH (J/mol)': [dH_12, dH_23, dH_31, dH_12 + dH_23 + dH_31]
})

print(f"Temperaturas: T1={T1:.1f} K | T2={T2:.1f} K | T3={T3:.2f} K\n")
print(tabela.to_string(index=False, float_format=lambda x: f"{x:9.2f}"))
```
