```python
!pip install CoolProp -q
from CoolProp . CoolProp import PropsSI

# Exemplo: Vapor umido de agua a 2 bar (200 kPa) com entalpia de 1800 kJ/kg
P_pa = 2.0 * 100000.0
h_mistura = 1800.0 * 1000.0  # Conversao obrigatoria de kJ/kg para J/kg

hf_agua = PropsSI('H', 'P', P_pa, 'Q', 0, 'Water')
hg_agua = PropsSI('H', 'P', P_pa, 'Q', 1, 'Water')

# Verificacao fisica antes do calculo do titulo
if hf_agua <= h_mistura <= hg_agua:
    x = (h_mistura - hf_agua) / (hg_agua - hf_agua)
    print(f"Mistura bifasica confirmada. Titulo (x): {x:.4f} ({x*100:.1f}% vapor)")
else:
    print("Atencao: O fluido nao esta na regiao de saturacao!")
```
