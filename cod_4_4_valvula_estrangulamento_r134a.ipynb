```python
!pip install CoolProp -q
import CoolProp.CoolProp as CP

T1 = 40.0 + 273.15   # K (condensador)
T2 = -10.0 + 273.15  # K (evaporador)

P1 = CP.PropsSI('P', 'T', T1, 'Q', 0, 'R134a')
h1 = CP.PropsSI('H', 'T', T1, 'Q', 0, 'R134a')

P2 = CP.PropsSI('P', 'T', T2, 'Q', 0, 'R134a')
h2 = h1
x2 = CP.PropsSI('Q', 'P', P2, 'H', h2, 'R134a')

print(f"Pressao alta (P1): {P1/1e5:.2f} bar | Pressao baixa (P2): {P2/1e5:.2f} bar")
print(f"Entalpia (h1=h2):  {h1/1000:.2f} kJ/kg")
print(f"Titulo na saida da valvula (x2): {x2:.4f} ({x2*100:.2f}% vapor)")
```
