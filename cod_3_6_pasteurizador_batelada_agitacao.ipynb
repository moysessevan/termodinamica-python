```python
m_leite = 500.0       # kg
cp_leite = 3930.0     # J/(kg.K)
T1 = 4.0 + 273.15     # K
T2 = 65.0 + 273.15    # K
W_dot_s = 1200.0      # W (trabalho de eixo entra -> positivo)
dt = 30.0 * 60.0      # s

dU = m_leite * cp_leite * (T2 - T1)
Ws = W_dot_s * dt
Q = dU - Ws

print(f"Energia interna acumulada (dU): {dU/1e3:8.2f} kJ")
print(f"Trabalho de eixo (Ws):          {Ws/1e3:8.2f} kJ")
print(f"Calor pela camisa (Q):          {Q/1e3:8.2f} kJ")
```
