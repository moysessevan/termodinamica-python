```python
!pip install CoolProp --quiet
import CoolProp.CoolProp as CP

m = 2.0               # kg
P = 2e5               # Pa (2 bar)
x1 = 0.25             # Titulo inicial
T2 = 300.0 + 273.15   # K

# Estado 1: mistura bifasica (P, Q)
h1 = CP.PropsSI('H', 'P', P, 'Q', x1, 'Water')
u1 = CP.PropsSI('U', 'P', P, 'Q', x1, 'Water')
v1 = 1.0 / CP.PropsSI('D', 'P', P, 'Q', x1, 'Water')

# Estado 2: vapor superaquecido (P, T)
h2 = CP.PropsSI('H', 'P', P, 'T', T2, 'Water')
u2 = CP.PropsSI('U', 'P', P, 'T', T2, 'Water')
v2 = 1.0 / CP.PropsSI('D', 'P', P, 'T', T2, 'Water')

Wb = -m * P * (v2 - v1)
dU = m * (u2 - u1)
Q = m * (h2 - h1)

print(f"Trabalho de fronteira (Wb): {Wb/1000:.2f} kJ")
print(f"Variacao de energia interna (dU): {dU/1000:.2f} kJ")
print(f"Calor fornecido (Q): {Q/1000:.2f} kJ")
print(f"Verificacao (dU - Wb - Q): {(dU - Wb - Q):.6f} J")
```
