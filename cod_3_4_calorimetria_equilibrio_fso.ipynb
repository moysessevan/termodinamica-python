```python
!pip install CoolProp --quiet
import CoolProp.CoolProp as CP
from scipy.optimize import fsolve

P_atm = 101325.0
m_cu, cp_cu, T_cu1 = 2.0, 385.0, 150.0 + 273.15
m_ag, T_ag1 = 10.0, 25.0 + 273.15

u_ag1 = CP.PropsSI('U', 'T', T_ag1, 'P', P_atm, 'Water')

def residuo_energia(T_guess):
    T = T_guess[0]
    u_ag2 = CP.PropsSI('U', 'T', T, 'P', P_atm, 'Water')
    return m_cu * cp_cu * (T - T_cu1) + m_ag * (u_ag2 - u_ag1)

T_eq = fsolve(residuo_energia, [30.0 + 273.15])[0]
print(f"Temperatura de equilibrio: {T_eq - 273.15:.2f} C")
```
