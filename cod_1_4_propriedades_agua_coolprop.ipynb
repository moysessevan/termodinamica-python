```python
!pip install CoolProp -q
from CoolProp.CoolProp import PropsSI

# Caso (a): Agua liquida a 25 C (298.15 K) e 1 atm (101325 Pa)
T_a = 25.0 + 273.15
P_a = 101325.0

rho_a = PropsSI('D', 'P', P_a, 'T', T_a, 'Water')
h_a = PropsSI('H', 'P', P_a, 'T', T_a, 'Water')

print(f"(a) Densidade da agua: {rho_a:.2f} kg/m3")
print(f"(a) Entalpia especifica: {h_a / 1000:.2f} kJ/kg")

# Caso (b): Temperatura de saturacao a 20 kPa (20000 Pa)
# Na saturacao, fornecemos P e o titulo Q (0 para liquido saturado)
P_b = 20000.0
T_sat_K = PropsSI('T', 'P', P_b, 'Q', 0, 'Water')
T_sat_C = T_sat_K - 273.15

print(f"(b) Temperatura de ebulicao a 20 kPa: {T_sat_C:.2f} C ({T_sat_K:.2f} K)")
```
