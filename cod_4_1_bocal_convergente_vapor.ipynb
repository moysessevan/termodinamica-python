```python
!pip install CoolProp -q
import CoolProp.CoolProp as CP
import numpy as np

P1, T1 = 15e5, 300.0 + 273.15
P2, T2 = 8e5, 240.0 + 273.15
A2 = 12e-4  # m2 (12 cm2)

h1 = CP.PropsSI('H', 'P', P1, 'T', T1, 'Water')
h2 = CP.PropsSI('H', 'P', P2, 'T', T2, 'Water')
D2 = CP.PropsSI('D', 'P', P2, 'T', T2, 'Water')

vel_2 = np.sqrt(2.0 * (h1 - h2))
m_dot = D2 * vel_2 * A2

print(f"h1 = {h1/1000:.2f} kJ/kg | h2 = {h2/1000:.2f} kJ/kg")
print(f"Velocidade de saida (V2): {vel_2:.2f} m/s")
print(f"Vazao massica (m_dot):    {m_dot:.4f} kg/s")
```
