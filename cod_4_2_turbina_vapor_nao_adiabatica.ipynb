```python
!pip install CoolProp -q
import CoolProp.CoolProp as CP

m_dot = 10.0         # kg/s
Q_dot = -50.0e3      # W (calor perdido -> negativo)

h1 = CP.PropsSI('H', 'P', 40e5, 'T', 450.0 + 273.15, 'Water')
h2 = CP.PropsSI('H', 'P', 1e5, 'Q', 1.0, 'Water')

W_dot_s = m_dot * (h2 - h1) - Q_dot

print(f"Entalpia entrada (h1): {h1/1000:.2f} kJ/kg")
print(f"Entalpia saida   (h2): {h2/1000:.2f} kJ/kg")
print(f"Potencia gerada  (Ws): {W_dot_s/1e6:.3f} MW ({W_dot_s/1000:.1f} kW)")
```
