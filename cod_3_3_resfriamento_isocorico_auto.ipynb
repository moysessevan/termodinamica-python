```python
!pip install CoolProp --quiet
import CoolProp.CoolProp as CP

m = 0.75              # kg
P1 = 1200e3           # Pa
T1 = 300.0 + 273.15   # K
P2 = 300e3            # Pa

D1 = CP.PropsSI('D', 'P', P1, 'T', T1, 'Water')
u1 = CP.PropsSI('U', 'P', P1, 'T', T1, 'Water')

# Estado 2: mesmo volume especifico -> D2 = D1
D2 = D1
T2 = CP.PropsSI('T', 'P', P2, 'D', D2, 'Water')
u2 = CP.PropsSI('U', 'P', P2, 'D', D2, 'Water')
x2 = CP.PropsSI('Q', 'P', P2, 'D', D2, 'Water')
fase2 = CP.PhaseSI('P', P2, 'D', D2, 'Water')

Q = m * (u2 - u1)

print(f"Fase no estado 2: {fase2}")
print(f"T2 = {T2 - 273.15:.2f} C | Titulo x2 = {x2:.4f}")
print(f"Trabalho W = 0.00 kJ | Calor Q = {Q/1000:.2f} kJ")
```
