```python
# Comentários começam com '#' e são ignorados pelo Python durante a compilação.
# Boa prática: nomeie variáveis de forma clara e sem acentos.
temperatura_celsius = 25.0 # Tipo float (número real)
pressao_atm = 1.0       # Pressão atmosférica padrão
numero_estagios = 3     # Tipo int (número inteiro)
fluido = "Water"        # Tipo str (texto / string)

# Operações matemáticas básicas:
# Soma (+), Subtração (-), Multiplicação (*), Divisão (/), Potência (**)
temperatura_kelvin = temperatura_celsius + 273.15
pressao_pascal = pressao_atm * 101325.0
energia_cinetica_especifica = (10.0 ** 2) / 2.0  # v^2 / 2 em J/kg

# Exibindo os resultados formatados com f-strings (forma moderna e recomendada)
print(f"Fluido estudado: {fluido}")
print(f"Temperatura absoluta: {temperatura_kelvin:.2f} K")
print(f"Pressão no SI: {pressao_pascal:.1f} Pa")
```
